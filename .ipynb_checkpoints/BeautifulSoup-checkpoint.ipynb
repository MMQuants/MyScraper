```python
from bs4 import BeautifulSoup
import requests
import pandas as pd
import csv
```

```python
url = 'https://www.booking.com/searchresults.bg.html?ss=Varna+City%2C+Varna+Province%2C+Bulgaria&efdco=1&label=mkt123sc-6c610bbb-c160-400c-9891-f23d7b138636&aid=304142&lang=bg&sb=1&src_elem=sb&src=index&dest_id=-839465&dest_type=city&ac_meta=GiAwN2U4MjU3ZjgzYjg5M2I3NTA3NjhjODI2ZmRhMTk5YyAAKAEyAmVuOgV2YXJuYQ%3D%3D&ac_position=0&ac_click_type=b&ac_langcode=en&ac_suggestion_list_length=5&search_selected=true&search_pageview_id=07e8257f83b893b750768c826fda199c&checkin=2026-11-11&checkout=2026-11-12&group_adults=2&no_rooms=1&group_children=0'
headers = {
    'User-Agent': 'Mozilla/5.0 (X11; CrOS x86_64 8172.45.0) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/51.0.2704.64 Safari/537.36',
    'Accept-Language': 'en-US, en;q=0.5'
}

response  = requests.get(url, headers=headers)
```

```python
soup = BeautifulSoup(response.text, 'html.parser')
```

```python
# Find all the hotel elements in the HTML document
hotels = soup.find_all('div', {'data-testid': 'property-card'})

hotels_data = []
# Loop over the hotel elements and extract the desired data
for hotel in hotels:
    # Extract the hotel name
    name_element = hotel.find('div', {'data-testid': 'title'})
    name = name_element.text.strip()

    # Extract the hotel location
    location_element = hotel.find('span', {'data-testid': 'address'})
    location = location_element.text.strip()

    # Extract the hotel price
    price_element = hotel.find('span', {'data-testid': 'price-and-discounted-price'})
    price = price_element.text.strip()

    # Extract the hotel rating
    rating_element = hotel.find('div', {'class': 'b5cd09854e d10a6220b4'})
    rating = rating_element.text.strip()

    # Append hotes_data with info about hotel
    hotels_data.append({
        'name': name,
        'location': location,
        'price': price,
        'rating': rating
    })
```

```python

hotels = pd.DataFrame(hotels_data)
hotels.head()
```
